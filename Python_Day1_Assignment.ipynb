## 10. Number Exercises

```python
number = int(input("Check a number for primality: "))
if number < 2:
    print("Not a prime number")
else:
    is_prime = True
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            is_prime = False
            break
    print("Prime number" if is_prime else "Not a prime number")
```

```python
limit = int(input("Enter a positive limit: "))
even_total = 0
odd_total = 0
for value in range(1, limit + 1):
    if value % 2 == 0:
        even_total += value
    else:
        odd_total += value
print("Even total:", even_total)
print("Odd total:", odd_total)
```

```python
number = int(input("Enter a number for factorial: "))
if number < 0:
    print("Factorial is not defined for negative numbers")
else:
    factorial = 1
    for value in range(1, number + 1):
        factorial *= value
    print(f"{number}! =", factorial)
```
