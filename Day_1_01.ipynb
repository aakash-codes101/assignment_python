```python
N=int(input("Enter the number"))
sum_even=0
sum_odd=0
for i in range(1,N+1):
  if i % 2 ==0:
    sum_even +=i
  else:
    sum_odd +=i
print("sum of even numbers:",sum_even)
print("sum of odd numbers:",sum_odd)
```

```python
# Factorial of a number
num= int(input("Enter the number"))
fact=1
if num <0:
  print("false")
elif num==0:
  print("fact of 0 is 1")
else:
  for i in range (1,num + 1):
    fact*=i
  print(f"fact of {num} is {fact}")
```

```python
# Exploring list functions using a fruit list

fruits = ["Apple", "Banana", "Mango", "Orange"]

print(fruits)
print(len(fruits))

fruits.append("Grapes")

more_fruits = ["Pineapple", "Cherry"]
fruits.extend(more_fruits)

print("After extend:", fruits)

print("First fruit:", fruits[0])
print("Last fruit:", fruits[-1])

fruits[1] = "Strawberry"
print("After updating index 1:", fruits)

fruits.remove("Mango")
print("After remove Mango:", fruits)

removed_fruit = fruits.pop(2)
print("Removed fruit:", removed_fruit)
print("After pop:", fruits)

fruits.sort()
print("After sort:", fruits)

fruits.reverse()
print("After reverse:", fruits)

print("Index of Apple:", fruits.index("Apple"))

print("Count of Apple:", fruits.count("Apple"))

fruits.insert(1, "Watermelon")
print("After insert:", fruits)

fruits_copy = fruits.copy()
print("Copied list:", fruits_copy)
```

```python
# A dictionary is written as key-value pairs.
# A key and its value are separated by a colon :
print("DICTIONARY OPERATIONS")
my_dict = {'name': 'Alice', 'age': 30, 'city': 'New York'}
print("Original Dictionary:", my_dict)
# Accessing Values
print(my_dict['name'])
# print(my_dict['country'])  # This gives KeyError
print(my_dict.get('city'))
# get() is safer: returns None if key doesn't exist
print(my_dict.get('country'))       # None
print(my_dict.get('country', 'USA'))  # USA
# Add a new key-value pair
my_dict['email'] = 'alice@example.com'
print("After adding email:", my_dict)
# Update an existing value
my_dict['age'] = 31
print("After updating age:", my_dict)
```
